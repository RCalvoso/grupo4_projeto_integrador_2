```python
import os
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from google.colab import drive

# 1. Garantir que o Drive está montado
if not os.path.exists('/content/drive'):
    drive.mount('/content/drive')

# 2. Localizar o arquivo chamados.csv dinamicamente no Drive
file_path = None
for root, dirs, files in os.walk('/content/drive/MyDrive'):
    if 'chamados.csv' in files:
        file_path = os.path.join(root, 'chamados.csv')
        break

if file_path:
    print(f"✅ Arquivo localizado em: {file_path}")
    df = pd.read_csv(file_path)
    print(f"Dataset original: {df.shape[0]} linhas e {df.shape[1]} colunas.")

    # 3. Definição das colunas a serem descartadas (Alerta Crítico)
    cols_to_drop = [
        'last_event_date', 'event_types', 'num_images_y',
        'monthly_revenue', 'contract_type', 'employees', 'city_y',
        'segment', 'company', 'vip', 'customer_since', 'state_y'
    ]

    # Dropar colunas inviáveis
    cols_existentes = [col for col in cols_to_drop if col in df.columns]
    df_clean = df.drop(columns=cols_existentes)
    print(f"Dataset após limpeza de colunas inviáveis: {df_clean.shape[0]} linhas e {df_clean.shape[1]} colunas.")

    # 4. Tratamento de nulos em textos e caminhos de imagem
    df_clean['title'] = df_clean['title'].fillna('')
    df_clean['description'] = df_clean['description'].fillna('')
    df_clean['image_path'] = df_clean['image_path'].fillna('')

    # 5. Split Estratificado: Treino (70%), Validação (15%) e Teste (15%)
    train_df, temp_df = train_test_split(
        df_clean, test_size=0.30, random_state=42, stratify=df_clean['target_category']
    )

    val_df, test_df = train_test_split(
        temp_df, test_size=0.50, random_state=42, stratify=temp_df['target_category']
    )

    print(f"\n=== Divisão dos Dados ===")
    print(f"Treino:    {len(train_df)} amostras ({(len(train_df)/len(df_clean))*100:.0f}%)")
    print(f"Validação: {len(val_df)} amostras ({(len(val_df)/len(df_clean))*100:.0f}%)")
    print(f"Teste:     {len(test_df)} amostras ({(len(test_df)/len(df_clean))*100:.0f}%)")

    # 6. Salvar os conjuntos processados na pasta do projeto
    output_dir = os.path.dirname(file_path)
    train_df.to_csv(os.path.join(output_dir, 'train_clean.csv'), index=False)
    val_df.to_csv(os.path.join(output_dir, 'val_clean.csv'), index=False)
    test_df.to_csv(os.path.join(output_dir, 'test_clean.csv'), index=False)

    print("\n✅ Arquivos 'train_clean.csv', 'val_clean.csv' e 'test_clean.csv' gerados com sucesso!")
else:
    print("❌ O arquivo chamados.csv não foi encontrado no Google Drive.")
```
