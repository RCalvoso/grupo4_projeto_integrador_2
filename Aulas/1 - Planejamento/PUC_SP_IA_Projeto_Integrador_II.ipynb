Educação centrada no aluno:

> Coding Dojo (ou simplesmente Dojo) em programação é uma metodologia de prática colaborativa e hands-on onde um grupo de desenvolvedores se reúne para resolver problemas de programação de forma iterativa e disciplinada. O objetivo principal é melhorar as habilidades de codificação, compartilhar conhecimentos, praticar TDD (Test-Driven Development), refatoração e boas práticas de design de software de maneira descontraída, mas focada. Tipicamente, utiliza-se o formato Randori (duas pessoas codificando no projetor enquanto o grupo observa e participa) ou Kata (exercícios repetidos como formas de artes marciais, para aperfeiçoar técnicas específicas). É inspirado na filosofia das artes marciais japonesas, onde o “dojo” é o local de treino, promovendo uma cultura de aprendizado contínuo, feedback imediato e colaboração em vez de competição.
